```python
#Write a Python program to find out the top 15 complaint types. Make a bar graph of complaint counts as well.
import pandas as pd
import matplotlib.pyplot as plt

# Loaded the dataset hehe
df = pd.read_csv('Customer Service Requests.csv', low_memory=False)

# Convert date columns to datetime
df['Created Date'] = pd.to_datetime(df['Created Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')

# Calculate Request_Closing_Time in hours
df['Request_Closing_Time'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600

# Top 15 complaint types by count
top_complaints = df['Complaint Type'].value_counts().head(15)

# Check if top_complaints is empty
if top_complaints.empty:
    print("No data available for complaint types.")
else:
    # Plot a bar graph
    plt.figure(figsize=(10, 6))
    top_complaints.plot(kind='bar', color='lightblue')
    plt.title('Top 15 Complaint Types')
    plt.xlabel('Complaint Type')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    
    # Save the plot
    plt.savefig('top_15_complaint_types.png')
    
    # Display the plot
    plt.show()
    
    # Close the plot to free memory
    plt.close()
```

```python
# Which complaint type has the highest average resolution time? Illustrate with a bar graph

# Calculate the average Request_Closing_Time for each complaint type
avg_closing_time = df.groupby('Complaint Type')['Request_Closing_Time'].mean()

# Identify the complaint type with the highest average resolution time
max_complaint = avg_closing_time.idxmax()
max_time = avg_closing_time.max()

# Top 15 for visualization
top_avg_closing = avg_closing_time.sort_values(ascending=False).head(15)

# Plot a bar graph
plt.figure(figsize=(10, 6))
bars = plt.bar(top_avg_closing.index, top_avg_closing.values, color='lightblue')

# this color is for the highest average resolution time
bars[top_avg_closing.index.get_loc(max_complaint)].set_color('grey')

# Green line at the maximum value
plt.axhline(y=max_time, color='green', linestyle='--')
plt.title('Top 15 Complaint Types by Average Resolution Time (Hours)')
plt.xlabel('Complaint Type')
plt.ylabel('Average Resolution Time (Hours)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Save the plot (uncommented to ensure the file is saved)
plt.savefig('highest_resolution_time.png')

# Display the plot
plt.show()
```

```python
# Write a Python program to show a histogram and box plot of any chosen different variables. Use proper labels in the graph.

import pandas as pd
import matplotlib.pyplot as plt

# Define functions to plot histogram and box plot
def plot_histogram(data, title, xlabel, ylabel, filename):
    plt.figure(figsize=(8, 6))
    plt.hist(data.dropna(), bins=30, color='lightblue', edgecolor='black')  # Drop NaN values
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(filename)
    plt.show()  # Display the plot
    plt.close()  # Close after displaying

def plot_boxplot(data, title, ylabel, filename):
    plt.figure(figsize=(8, 6))
    plt.boxplot(data.dropna(), vert=True, patch_artist=True, boxprops=dict(facecolor='lightgreen'))  # Drop NaN values
    plt.title(title)
    plt.ylabel(ylabel)
    plt.xticks([1], ['Request_Closing_Time'])
    plt.tight_layout()
    plt.savefig(filename)  
    plt.show()  
    plt.close()  

# Plotting Histogram
plot_histogram(df['Request_Closing_Time'], 'Histogram of Request Closing Time', 'Request Closing Time (Hours)', 'Frequency', 'closing_time_histogram.png')

# Plotting Box Plot
plot_boxplot(df['Request_Closing_Time'], 'Box Plot of Request Closing Time', 'Request Closing Time (Hours)', 'closing_time_boxplot.png')
```

```python

import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Customer Service Requests.csv', low_memory=False)

# Convert dates and calculate resolution time in hours
df['Created Date'] = pd.to_datetime(df['Created Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Resolution_Hours'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600

# Keeping only necessary columns
keep_columns = ['Complaint Type', 'Borough', 'Resolution_Hours']
df = df[keep_columns].dropna()

# Get average resolution time by borough and complaint type
avg_times = df.groupby(['Borough', 'Complaint Type'])['Resolution_Hours'].mean().unstack()

# Select top 5 complaint types by frequency
top_complaints = df['Complaint Type'].value_counts().head(5).index
avg_times = avg_times[top_complaints]

# Plot bar chart
if not avg_times.empty:
    plt.figure(figsize=(12, 6))
    avg_times.plot(kind='bar', colormap='Set2')
    plt.title('Average Complaint Resolution Time by Borough and Complaint Type')
    plt.xlabel('Borough')
    plt.ylabel('Average Resolution Time (Hours)')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Complaint Type', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('complaint_resolution_by_borough.png')
    plt.show()
    plt.close()
else:
    print("No data available for plotting.")
    
    
    
    
    
    
    
```

```python
import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv('Customer Service Requests.csv', low_memory=False)

# Convert dates and calculate resolution time in hours
df['Created Date'] = pd.to_datetime(df['Created Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df['Resolution_Hours'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600

# Keeping only necessary columns
df = df[['Complaint Type', 'Resolution_Hours']].dropna()

# Group resolution times by complaint type
groups = [df[df['Complaint Type'] == complaint]['Resolution_Hours'] 
          for complaint in df['Complaint Type'].unique()]

# Run ANOVA test and provide p-value
if len(groups) > 1:
    f_stat, p_value = f_oneway(*groups)
    print("ANOVA Test: Resolution Time by Complaint Type")
    print(f"F-Statistic: {f_stat:.2f}, P-Value: {p_value:.4f}")
    # Interpret results to accept or reject Null Hypothesis
    if p_value < 0.05:
        print("Result: Null hypothesis rejected. Resolution times differ across complaint types.")
    else:
        print("Result: Null hypothesis accepted. Resolution times are similar across complaint types.")
else:
    print("Not enough complaint types for ANOVA test.")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load dataset
df = pd.read_csv('Customer Service Requests.csv', low_memory=False)

# Keep only necessary columns
df = df[['Complaint Type', 'Borough']].dropna()

# Create contingency table
table = pd.crosstab(df['Complaint Type'], df['Borough'])

# Run Chi-Square test and provide p-value
if table.size > 0:
    chi2_stat, p_value, dof, _ = chi2_contingency(table)
    print("Chi-Square Test: Complaint Type vs Borough")
    print(f"Chi-Square: {chi2_stat:.2f}, P-Value: {p_value:.4f}, Degrees of Freedom: {dof}")
    # Interpret results to accept or reject Null Hypothesis
    if p_value < 0.05:
        print("Result: Null hypothesis rejected. Complaint Type and Borough are related.")
    else:
        print("Result: Null hypothesis accepted. Complaint Type and Borough are not related.")
else:
    print("No data for Chi-Square test.")
```
